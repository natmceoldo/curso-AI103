# Guia de Instalação: Python e VS Code para Notebooks (`.ipynb`)

Este guia apresenta o passo a passo completo para configurar um ambiente do zero capaz de criar e executar arquivos Jupyter Notebook (`.ipynb`) no Visual Studio Code.

### 1. Instalar o Python

1. Acesse o site oficial: [python.org/downloads](https://www.python.org/downloads/).
2. Baixe o instalador recomendado para o seu sistema operacional.
3. Execute o instalador:
   * **Importante (Windows):** Na primeira tela, marque a opção **"Add python.exe to PATH"** antes de prosseguir.
   * Clique em **Install Now** e aguarde a conclusão.
4. Para confirmar a instalação, abra o terminal do sistema (Prompt de Comando, PowerShell ou Terminal) e execute:
   ```bash
   python --version
   ```
   *(No Windows, caso `python` não seja reconhecido imediatamente, teste `py --version`).*

### 2. Instalar o Visual Studio Code (VS Code)

1. Acesse o site oficial: [code.visualstudio.com](https://code.visualstudio.com/).
2. Baixe e execute o instalador para o seu sistema.
3. Siga as etapas padrão de instalação e abra o VS Code ao finalizar.


### 3. Instalar as Extensões Obrigatórias no VS Code

O VS Code renderiza textos em Markdown nativamente, mas precisa de extensões oficiais para reconhecer o Python e executar células de código em notebooks.

1. No VS Code, abra a aba de **Extensões** usando o atalho `Ctrl + Shift + X` (ou `Cmd + Shift + X` no macOS).
2. Pesquise e instale as duas extensões abaixo (ambas publicadas pela **Microsoft**):
   * **Python** (identificador: `ms-python.python`)
   * **Jupyter** (identificador: `ms-toolsai.jupyter`)
3. Caso o VS Code solicite, clique em **Reload Required** para recarregar o editor.

### 4. Instalar o Pacote `ipykernel`

O `ipykernel` é a biblioteca responsável por conectar o código Python à interface do notebook.

1. Abra uma pasta de trabalho no VS Code (`File > Open Folder...`). Caso apareça o aviso **Do you trust the authors of the files in this folder?**, clique em **Yes, I trust the authors** (o modo restrito bloqueia a execução de código).
2. Abra o terminal integrado do VS Code com o atalho ``Ctrl + ` `` (ou `Ctrl + '`).
3. Execute os comandos abaixo para atualizar o gerenciador de pacotes e instalar o kernel:
   ```bash
   python -m pip install --upgrade pip
   python -m pip install ipykernel
   ```
4. *(Opcional)* Registre o kernel explicitamente no perfil do usuário:
   ```bash
   python -m ipykernel install --user
   ```


### 5. Criar e Executar seu Primeiro Notebook (`.ipynb`)


1. No explorador de arquivos do VS Code, crie um novo arquivo com a extensão `.ipynb` (exemplo: `teste.ipynb`).
2. No canto superior direito do editor, clique em **Select Kernel** (*Selecionar Kernel*).
3. Selecione **Python Environments...** e escolha a versão do Python instalada no Passo 1.
4. Adicione uma célula de código (`+ Code`), digite um comando simples:
   ```python
   print("Ambiente configurado com sucesso!")
   10 + 5
   ```
5. Pressione `Shift + Enter` (ou clique no ícone de **Play** ao lado da célula) para executar.

### 6. Solução Rápida de Problemas


* **O Python não aparece na lista de Kernels:** Pressione `Ctrl + Shift + P`, digite `Developer: Reload Window` e pressione `Enter` para forçar o VS Code a reescanear os ambientes instalados.
* **Select Kernel fica carregando infinitamente:** Verifique se o VS Code e a extensão **Jupyter** estão atualizados. Versões antigas da extensão podem falhar ao iniciar em builds recentes do VS Code (confira os detalhes em `Ctrl + Shift + U` > aba **Jupyter**).
* **Uso de Ambientes Virtuais (`venv`):** Caso utilize um ambiente virtual no projeto, ative-o no terminal antes de rodar `pip install ipykernel` e selecione o Python de dentro da pasta `.venv` em **Select Kernel**.
